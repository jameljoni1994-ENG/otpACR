# 00 — Baseline Cubic Regularization (CR)

Exact Nesterov cubic regularization on analytic problems. Compare optimality gap $f(x_k)-f^*$ vs iterations.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import matplotlib.pyplot as plt
from cubic_reg.problems import Quadratic, Rosenbrock
from cubic_reg.solvers import cr

%matplotlib inline
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
for kappa in [10, 100, 1000]:
    p = Quadratic(n=50, condition=float(kappa), seed=0)
    res = cr.minimize(p, x0=np.ones(p.dim), M=1.0, eps=1e-10, max_iter=80)
    gap = np.maximum(np.asarray(res.history_f) - p.f_star, 1e-16)
    ax.semilogy(gap, label=f"kappa={kappa}, nit={res.nit}")
ax.set_xlabel("iteration")
ax.set_ylabel(r"$f(x_k) - f^*$")
ax.set_title("CR on quadratics")
ax.legend()
ax.grid(True, which="both", alpha=0.3)
plt.show()
```

```python
p = Rosenbrock(n=2)
res = cr.minimize(p, x0=np.array([-1.2, 1.0]), M=10.0, eps=1e-6, max_iter=200)
print(res.success, res.nit, res.grad_norm, res.f, res.message)
plt.figure(figsize=(6, 3))
plt.semilogy(res.history_grad_norm)
plt.xlabel("iteration")
plt.ylabel(r"$\|\nabla f\|$")
plt.title("CR on Rosenbrock")
plt.grid(True, which="both", alpha=0.3)
plt.show()
```
