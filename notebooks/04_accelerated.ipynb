# 04 — Accelerated Cubic Regularization

Test whether accelerated CR reduces $f(x_k)-f^*$ faster than baseline CR on well-conditioned and ill-conditioned quadratics.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import matplotlib.pyplot as plt
from cubic_reg.problems import Quadratic
from cubic_reg.solvers import cr, accelerated
from cubic_reg.plotting import plot_optimality_gap

%matplotlib inline
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5), sharey=True)
for ax, kappa in zip(axes, [10, 100, 1000]):
    p = Quadratic(n=40, condition=float(kappa), seed=0)
    x0 = np.ones(p.dim)
    results = {
        "CR": cr.minimize(p, x0=x0, M=1.0, eps=1e-10, max_iter=120),
        "Acc-CR": accelerated.minimize(p, x0=x0, M=1.0, eps=1e-10, max_iter=120),
    }
    plot_optimality_gap(results, f_star=p.f_star, ax=ax, title=f"kappa={kappa}")
plt.tight_layout(); plt.show()
```

Reference slopes: baseline CR often behaves like $O(1/k^2)$; accelerated theory predicts $O(1/k^3)$ on smooth convex problems.
